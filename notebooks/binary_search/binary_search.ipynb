## Core Concepts

Binary search eliminates half of the current search space at each step, giving it `O(log n)` time complexity. It is not limited to finding a value in a sorted array: it can apply whenever a monotonic condition tells us which half of a search space to discard.

This Notebook uses the closed interval `[left, right]`: if the target exists, it must remain inside that interval. The matching rules are `left = 0`, `right = len(nums) - 1`, and the loop condition `left <= right`. After checking `mid`, exclude it with `mid - 1` or `mid + 1` when it is not the answer. Do not mix closed-interval and half-open-interval rules.


```python
def binary_search_template(nums, target):
    """Return the index of target in an ascending array, or -1 if absent."""
    left, right = 0, len(nums) - 1

    while left <= right:
        mid = left + (right - left) // 2
        if nums[mid] == target:
            return mid
        if nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```

## Problem 01 LeetCode 704. Binary Search — Easy


Given an ascending array of distinct integers and a target, return the target's index if it exists. Otherwise, return `-1`. The required time complexity is `O(log n)`.


```python
from typing import List


class BinarySearchSolution:
    def search(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = left + (right - left) // 2

            if nums[mid] == target:
                return mid
            if nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        return -1
```

## Problem 04 LeetCode 69. Sqrt(x) — Easy


Given a non-negative integer `x`, return the square root of `x` rounded down to the nearest integer. Do not use a built-in exponent function or operator.


```python
class SqrtSolution:
    def my_sqrt(self, x: int) -> int:
        if x < 2:
            return x

        left, right = 1, x // 2

        while left <= right:
            mid = (left + right) // 2
            mid_squared = mid * mid

            if mid_squared == x:
                return mid
            elif mid_squared > x:
                right = mid - 1
            else:
                left = mid + 1
        return right
```

```python
if __name__ == "__main__":
    solution = SqrtSolution()
    print(solution.my_sqrt(10))
```

## Problem 05 LeetCode 441. Arranging Coins — Easy


Given `n` coins, build a staircase whose `k`th row requires exactly `k` coins. Return the number of complete rows that can be formed.


```python
class ArrangeCoinsSolution:
    def arrange_coins(self, n: int) -> int:
        left, right = 1, n

        while left <= right:
            mid = (left + right) // 2
            coins_needed = (mid * (mid + 1)) // 2
            if coins_needed == n:
                return mid
            elif coins_needed > n:
                right = mid - 1
            else:
                left = mid + 1
        return right

    def arrange_coins1(self, n: int) -> int:
        left , right = 1, n + 1

        while left < right:
            mid = (left + right) // 2
            coins_needed = (mid * (mid + 1)) // 2

            if coins_needed == n:
                return mid
            elif coins_needed < n:
                left = mid + 1
            else:
                right = mid
        return right - 1
```

## Problem 06 LeetCode 34. Find First and Last Position of Element in Sorted Array — Medium


Given a non-decreasing integer array and a target, return the first and last indices at which the target appears. Return `[-1, -1]` if it is absent. The required time complexity is `O(log n)`.


```python
from typing import List


class FindFirstAndLastPositionSolution:
    def search_range(self, nums: List[int], target: int) -> List[int]:
        def left_bound() -> int:
            left, right = 0, len(nums)
            while left < right:
                mid = (left + right) // 2

                if nums[mid] < target:
                    left = mid + 1
                else:
                    right = mid
            return left

        def right_bound() -> int:
            left, right = 0, len(nums)
            while left < right:
                mid = (left + right) // 2
                if nums[mid] <= target:
                    left = mid + 1
                else:
                    right = mid
            return left - 1

        left, right = left_bound(), right_bound()
        if left < len(nums) and nums[left] == target:
            return [left, right]

        return [-1, -1]
```

```python
if __name__ == "__main__":
    solution = FindFirstAndLastPositionSolution()

    test_cases = [
        ([5, 7, 7, 8, 8, 10], 8, [3, 4]),  # Target appears twice.
        ([5, 7, 7, 8, 8, 10], 6, [-1, -1]),  # Target is absent.
        ([], 0, [-1, -1]),  # Empty array.
        ([1], 1, [0, 0]),  # Single element matches.
        ([1], 0, [-1, -1]),  # Single element does not match.
        ([2, 2], 2, [0, 1]),  # Every element matches.
        ([1, 2, 3, 3, 3, 4, 5], 3, [2, 4]),  # Target appears many times.
        ([1, 2, 3], 1, [0, 0]),  # Target is at the left boundary.
        ([1, 2, 3], 3, [2, 2]),  # Target is at the right boundary.
    ]

    for nums, target, expected in test_cases:
        actual = solution.search_range(nums, target)
        assert actual == expected, (
            f"search_range({nums}, {target}) returned {actual}; "
            f"expected {expected}"
        )

    print(f"✅ All {len(test_cases)} tests passed")
```

## Problem 07 LeetCode 744. Find Smallest Letter Greater Than Target — Easy


Given a non-decreasing array of lowercase letters and a target, return the smallest letter strictly greater than the target. If none exists, return the first letter.


```python
from typing import List


class NextGreatestLetterSolution:
    def next_greatest_letter(self, letters: List[str], target: str) -> str:
        left, right = 0, len(letters)

        if letters[right - 1] <= target:
            return letters[left]

        while left < right:
            mid = (left + right) // 2
            if letters[mid] <= target:
                left = mid + 1
            else:
                right = mid
        return letters[left]
```

```python
if __name__ == "__main__":
    solution = NextGreatestLetterSolution()

    test_cases = [
        (["c", "f", "j"], "a", "c"),  # Target is smaller than every letter.
        (["c", "f", "j"], "c", "f"),  # The answer must be strictly greater.
        (["c", "f", "j"], "d", "f"),  # Answer is between two letters.
        (["c", "f", "j"], "g", "j"),  # Answer is the last letter.
        (["c", "f", "j"], "j", "c"),  # Wrap around after an exact match.
        (["c", "f", "j"], "z", "c"),  # Wrap around above the largest letter.
        (["x", "x", "y", "y"], "x", "y"),  # Duplicate letters.
        (["e", "e", "e", "k", "q", "q", "v", "y"], "q", "v"),
    ]

    for letters, target, expected in test_cases:
        actual = solution.next_greatest_letter(letters, target)
        assert actual == expected, (
            f"next_greatest_letter({letters}, '{target}') returned '{actual}'; "
            f"expected '{expected}'"
        )

    print(f"✅ All {len(test_cases)} tests passed")
```

## Problem 08 LeetCode 1539. Kth Missing Positive Number — Easy


Given a strictly increasing array of positive integers and an integer `k`, return the `k`th positive integer missing from the array.


```python
from typing import List


class KthMissingPositiveSolution:
    def find_kth_positive(self, arr: List[int], k: int) -> int:
        left, right = 0, len(arr)

        while left < right:
            mid = (left + right) // 2
            mid_missing = arr[mid] - mid - 1

            if mid_missing < k:
                left = mid + 1
            else:
                right = mid
        return left + k
```

## Problem 09 LeetCode 540. Single Element in a Sorted Array — Medium


Every element in a sorted array appears exactly twice except for one element. Return that single element in `O(log n)` time and `O(1)` space.


```python
from typing import List


class SingleElementSortedArraySolution:
    def single_non_duplicate(self, nums: List[int]) -> int:
        left, right = 0, len(nums) - 1

        while left < right:
            mid = (left + right) // 2

            if mid % 2 == 1:
                mid -= 1
            if nums[mid] == nums[mid + 1]:
                left = mid + 2
            else:
                right = mid
        return nums[left]
```

```python
if __name__ == "__main__":
    solution = SingleElementSortedArraySolution()
    test_cases = [
        ([1, 1, 2, 3, 3, 4, 4, 8, 8], 2),
        ([3, 3, 7, 7, 10, 11, 11], 10),
        ([1], 1),
        ([1, 1, 2], 2),
        ([1, 2, 2], 1),
        ([1, 1, 2, 2, 3], 3),
    ]
    for nums, expected in test_cases:
        actual = solution.single_non_duplicate(nums)
        assert actual == expected, f"single_non_duplicate({nums}) returned {actual}; expected {expected}"
    print(f"✅ All {len(test_cases)} tests passed")
```

## Problem 10 LeetCode 33. Search in Rotated Sorted Array — Medium


Search for a target in a sorted array that may have been rotated at an unknown pivot. All values are distinct, and the required time complexity is `O(log n)`.


```python
from typing import List


class SearchRotatedArraySolution:
    def search(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target:
                return mid
            elif nums[left] < nums[mid]:
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
            else:
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
        return -1
```

```python
if __name__ == "__main__":
    solution = SearchRotatedArraySolution()
    test_cases = [
        ([4, 5, 6, 7, 0, 1, 2], 0, 4),
        ([4, 5, 6, 7, 0, 1, 2], 3, -1),
        ([1], 0, -1),
        ([1], 1, 0),
        ([1, 3], 3, 1),
        ([5, 1, 3], 5, 0),
        ([1, 2, 3, 4, 5], 4, 3),
    ]
    for nums, target, expected in test_cases:
        actual = solution.search(nums, target)
        assert actual == expected, f"search({nums}, {target}) returned {actual}; expected {expected}"
    print(f"✅ All {len(test_cases)} tests passed")
```

## Problem 11 LeetCode 81. Search in Rotated Sorted Array II — Medium


Determine whether a target exists in a rotated non-decreasing array that may contain duplicate values.


```python
from typing import List


class SearchRotatedArrayIISolution:
    def search(self, nums: List[int], target: int) -> bool:
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return True
            
            if nums[left] == nums[mid] == nums[right]:
                left += 1
                right -= 1
                continue


            # 为什么是 <= mid
            if nums[left] <= nums[mid]: 
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
            else:
                if nums[mid] <= target < nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1

        return False
```

## Problem 15 LeetCode 74. Search a 2D Matrix — Medium


Search a matrix whose rows are sorted and whose first value in each row is greater than the last value in the previous row. Return whether the target exists.


```python
from typing import List


class SearchMatrixSolution:
    def search_matrix(self, matrix: List[List[int]], target: int) -> bool:
        m = len(matrix)
        n = len(matrix[0])

        left, right = 0, m * n - 1

        while left <= right:
            mid = (left + right) // 2
            row = mid // n
            col = mid % n
            mid_value = matrix[row][col]

            if mid_value == target:
                return True
            if mid_value < target:
                left = mid + 1
            else:
                right = mid - 1
        return False
```

```python
if __name__ == "__main__":
    solution = SearchMatrixSolution()
    test_cases = [
        ([[1, 3, 5, 7], [10, 11, 16, 20], [23, 30, 34, 60]], 3, True),
        ([[1, 3, 5, 7], [10, 11, 16, 20], [23, 30, 34, 60]], 13, False),
        ([[1]], 1, True),
        ([[1]], 0, False),
        ([[1, 3, 5]], 5, True),
        ([[1], [3], [5]], 3, True),
        ([[1, 2], [4, 5]], 6, False),
    ]
    for matrix, target, expected in test_cases:
        actual = solution.search_matrix(matrix, target)
        assert actual is expected, f"search_matrix({matrix}, {target}) returned {actual}; expected {expected}"
    print(f"✅ All {len(test_cases)} tests passed")
```

## Problem 16 LeetCode 875. Koko Eating Bananas — Medium


Return the minimum integer eating speed that allows all banana piles to be finished within `h` hours.


```python
from typing import List


class KokoEatingBananasSolution:
    def min_eating_speed(self, piles: List[int], h: int) -> int:

        def can_finish(speed:int) -> bool:
            total = 0

            for pile in piles:
                total += ((pile + speed - 1) // speed)
            return total <= h

        left, right = 1, max(piles)

        while left < right:
            mid = (left + right) // 2
            if can_finish(mid):
                right = mid
            else:
                left = mid + 1
        return left
```

```python
if __name__ == "__main__":
    solution = KokoEatingBananasSolution()
    test_cases = [
        ([3, 6, 7, 11], 8, 4),
        ([30, 11, 23, 4, 20], 5, 30),
        ([30, 11, 23, 4, 20], 6, 23),
        ([1], 1, 1),
        ([1, 1, 1, 1], 4, 1),
        ([9, 9, 9], 6, 5),
    ]
    for piles, h, expected in test_cases:
        actual = solution.min_eating_speed(piles, h)
        assert actual == expected, f"min_eating_speed({piles}, {h}) returned {actual}; expected {expected}"
    print(f"✅ All {len(test_cases)} tests passed")
```
